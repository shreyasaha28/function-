```python
#4. Write a python function that checks if a given number is prime or not from 1 to 200.
```

```python
def is_prime(n):
    if n <= 1:
        return False  # 0 and 1 are not prime numbers
    if n == 2:
        return True   # 2 is the only even prime number
    if n % 2 == 0:
        return False  # Exclude even numbers greater than 2
    for i in range(3, int(n ** 0.5) + 1, 2):  # Check odd divisors only
        if n % i == 0:
            return False
    return True
```

```python
for i in range(1, 201):
    if is_prime(i):
        print(i, "is a prime number")
```

```python
#6. Write a generator function in python that yields the power of 2 upto a given component.
```

```python
def power_of_two(x):
    """Generator that yields powers of 2 from 2^0 up to 2^limit."""
    for i in range(x + 1):
        yield i ** 2
```

```python
for number in power_of_two(5):
    print(number)
```

```python
#9. Write a python program that uses map () to convert a list of trmperature from celcious to farenheit.
```

```python
celsius_temps = [0, 10, 20]

def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

fahrenheit_temps = list(map(celsius_to_fahrenheit, celsius_temps))
print(fahrenheit_temps)
```

```python
# 10.Write a python program that uses filter () to remove all the vovels from a given string.
```

```python
def is_not_vowel(char):
    vowels = "AEIOU"
    return char not in vowels

input_string = "AFRICA"

filtered_string = ''.join(filter(is_not_vowel, input_string))

print(filtered_string)
```

```python
#11.  Imagine an accounting routine used in a book shop. It works on a list with sublists, which look like this:
      order number           book title and author                                  quantity                 price per item
      34587                  learning python ,mark lutz                             4                           40.95
      98762                  programming python,mark lutz                           5                           56.80
      77226                  head first python, paul barry                          3                           32.95
      88112                  einufuhrung i python 3, bernd kelin                    3                           24.99



Write a Python program, which returns a list with 2-tuples. Each tuple consists of the order number and the
product of the price per item and the quantity. The product should be increased by 10,- € if the value of the
order is smaller than 100,00 €.

Write a Python program using lambda and map.
```

```python
orders = [
    [34587, "Learning Python, Mark Lutz", 4, 40.95],
    [98762, "Programming Python, Mark Lutz", 5, 56.80],
    [77226, "Head First Python, Paul Barry", 3, 32.95],
    [88112, "Einführung in Python 3, Bernd Klein", 3, 24.99]
]

order_totals = list(map(
    lambda order: (
        order[0],
        round(order[2] * order[3] + (10 if order[2] * order[3] < 100 else 0), 2)
    ),
    orders
))
```

```python
print(order_totals)
```
